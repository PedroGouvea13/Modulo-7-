```python
import pandas as pd
```

```python
#Funcao para Calcular o cubo de um numero
def eleva_cubo(x):
  return x ** 3
```

```python
#Funcao de lambda para elevar ao cubo
eleva_cubo_lambda = lambda x : x **3
print(eleva_cubo_lambda(3))
print(eleva_cubo(3))
```

```python
df = pd.DataFrame({'numeros': [1, 2, 3, 4, 5, 10]})
df['cubo_funcao'] = df['numeros'].apply(eleva_cubo)
df['cubo_lambda'] = df['numeros'].apply(eleva_cubo_lambda)
df
print(df)
```
